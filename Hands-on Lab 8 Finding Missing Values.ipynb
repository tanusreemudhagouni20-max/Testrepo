### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
missing_values = df.isnull().sum()
print("Missing Values per Column:\n")
print(missing_values)
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Write your code here
import seaborn as sns
import matplotlib.pyplot as plt

# Create a heatmap of missing values
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Heatmap of Missing Values in the Dataset")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Write your code here
column_name = "Employment"
missing_count = df[column_name].isnull().sum()

print(f"Number of missing rows in column '{column_name}': {missing_count}")
```

### 5. Visualizing Imputed Data

```python
## Write your code here

import matplotlib.pyplot as plt
import seaborn as sns

column_name = "Employment"

plt.figure(figsize=(10, 6))

sns.countplot(data=df, x=column_name, order=df[column_name].value_counts().index, palette="Set2")

plt.title(f"Distribution of '{column_name}' After Imputation")
plt.xlabel(column_name)
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```
