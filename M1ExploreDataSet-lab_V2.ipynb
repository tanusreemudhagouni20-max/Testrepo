Print the mean age of the survey participants.


```python
df['Age'] = df['Age'].astype(str).str.extract('(\d+)').astype(float)
mean_age = df['Age'].mean()
print("Mean age of survey participants:", mean_age)
```

```python
print(df.columns)
print(df['Age'].head(20))
print(df['Age'].dtype)

```

```python
mean_age = df['Age'].mean()

if pd.isna(mean_age):
    print("Mean age of survey participants cannot be calculated because age data is missing.")
else:
    print("Mean age of survey participants:", mean_age)
```
