#### Step 1: Import the necessary module.


### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
print("Column Data Types:")
print(df.dtypes)
print("\n")

print("Non-null Counts:")
print(df.count())
print("\n")

print("Missing Values:")
print(df.isnull().sum())
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
numerical_stats = df.describe()
print(numerical_stats)
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
unique_countries = df['Country'].unique()
print("Unique entries in Country column:")
print(unique_countries)
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
print("Unique countries:", df['Country'].unique())
print("\nUnique education levels:", df['EdLevel'].unique())
country_map = {
    'usa': 'United States',
    'US': 'United States',
    'U.S.': 'United States',
    'Untied States': 'United States',
    'uk': 'United Kingdom',
    'England': 'United Kingdom',
    'india ': 'India', 
}

edlevel_map = {
    'Bachelor’s degree': 'Bachelors',
    'Bachelors degree': 'Bachelors',
    'Master’s degree': 'Masters',
    'Master degree': 'Masters',
    'Some college/university study without earning a degree': 'Some College',
}
df['Country'] = df['Country'].str.strip().replace(country_map)
df['EdLevel'] = df['EdLevel'].str.strip().replace(edlevel_map)

print("Unique countries after standardization:", df['Country'].unique())
print("Unique education levels after standardization:", df['EdLevel'].unique())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
employment_dummies = pd.get_dummies(df['Employment'], prefix='Employment')

df = pd.concat([df, employment_dummies], axis=1)

print(df.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
missing_counts = df.isnull().sum()

missing_counts_sorted = missing_counts.sort_values(ascending=False)

print("Columns with the highest number of missing values:")
print(missing_counts_sorted.head(10)) 
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
median_salary = df['ConvertedCompYearly'].median()
df['ConvertedCompYearly'].fillna(median_salary, inplace=True)

print(f"Missing values in ConvertedCompYearly after imputation: {df['ConvertedCompYearly'].isnull().sum()}")
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_cols:
    median_val = df[col].median()
    df[col].fillna(median_val, inplace=True)

print("Missing values after imputation (numerical columns):")
print(df[numerical_cols].isnull().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
mode_remote = df['RemoteWork'].mode()[0]

df['RemoteWork'].fillna(mode_remote, inplace=True)

print(f"Missing values in RemoteWork after imputation: {df['RemoteWork'].isnull().sum()}")
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    mode_val = df[col].mode()[0]
    df[col].fillna(mode_val, inplace=True)

print("Missing values after imputation (categorical columns):")
print(df[categorical_cols].isnull().sum())

```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - df['ConvertedCompYearly'].min()) / \
                                   (df['ConvertedCompYearly'].max() - df['ConvertedCompYearly'].min())

print(df[['ConvertedCompYearly', 'ConvertedCompYearly_MinMax']].head())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])

print(df[['ConvertedCompYearly', 'ConvertedCompYearly_Log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
def categorize_experience(years):
    if years <= 2:
        return 'Beginner'
    elif years <= 5:
        return 'Intermediate'
    elif years <= 10:
        return 'Experienced'
    else:
        return 'Expert'

df['ExperienceLevel'] = df['YearsCodePro'].apply(categorize_experience)

print(df[['YearsCodePro', 'ExperienceLevel']].head())
```
